```python
import pandas as pd

df_ler_dados=pd.read_csv("/content/vendas_fevereiro.csv")
df_ler_dados.head()
```

```python
df_ler_dados1 =pd.read_csv("/content/vendas_janeiro.csv")
df_ler_dados1.head()
```

```python
df_ler_dados2 =pd.read_csv("/content/vendas_marco.csv")
df_ler_dados2.head()

```

```python
df_dados_juntos = pd.concat([df_ler_dados,df_ler_dados1,df_ler_dados2], ignore_index=True)
df_dados_juntos.to_csv('dadosJuntos.csv', index=False)
display(df_dados_juntos)
```

```python
from IPython.display import display
display(df_dados_juntos)
```

```python
display(df_dados_juntos.describe())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate the total sale amount for each row
df_dados_juntos['Total_Venda'] = df_dados_juntos['Preço'] * df_dados_juntos['Qtd']

# Group by 'Loja' and sum the 'Total_Venda'
sales_by_store = df_dados_juntos.groupby('Loja')['Total_Venda'].sum().reset_index()

# Create the bar chart
plt.figure(figsize=(10, 6))
sns.barplot(x='Loja', y='Total_Venda', data=sales_by_store, palette='viridis')
plt.title('Total de Vendas por Loja')
plt.xlabel('Loja')
plt.ylabel('Total de Vendas')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
# Group by 'Produto' and calculate the average of 'Total_Venda'
average_sales_by_product = df_dados_juntos.groupby('Produto')['Total_Venda'].mean().reset_index()

# Sort for better visualization (optional)
average_sales_by_product = average_sales_by_product.sort_values(by='Total_Venda', ascending=False)

print("Média de Vendas por Produto:")
display(average_sales_by_product)
```

```python
# Create the bar chart for average sales by product
plt.figure(figsize=(12, 7))
sns.barplot(x='Produto', y='Total_Venda', data=average_sales_by_product, palette='magma')
plt.title('Média de Vendas por Produto')
plt.xlabel('Produto')
plt.ylabel('Média de Vendas')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
